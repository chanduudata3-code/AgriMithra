```python
import os

JOWAR_DATASET_PATH = r"D:\AgriMithra datasets\Sorghum Disease Image Dataset\Sorghum Disease Image Dataset"

print("Dataset exists:", os.path.exists(JOWAR_DATASET_PATH))

print("\nJowar classes:")
for folder in sorted(os.listdir(JOWAR_DATASET_PATH)):
    print(folder)
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

jowar_train_ds = tf.keras.utils.image_dataset_from_directory(
    JOWAR_DATASET_PATH,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

jowar_val_ds = tf.keras.utils.image_dataset_from_directory(
    JOWAR_DATASET_PATH,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

# IMPORTANT: save class names BEFORE prefetching
jowar_class_names = jowar_train_ds.class_names

print("\nJowar classes:", jowar_class_names)
print("Number of classes:", len(jowar_class_names))
```

```python
AUTOTUNE = tf.data.AUTOTUNE

jowar_train_ds = jowar_train_ds.cache().shuffle(1000).prefetch(
    buffer_size=AUTOTUNE
)

jowar_val_ds = jowar_val_ds.cache().prefetch(
    buffer_size=AUTOTUNE
)

print("Jowar dataset optimization completed!")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

IMG_SIZE = (224, 224)

jowar_classes = [
    'Anthracnose and Red Rot',
    'Cereal Grain molds',
    'Covered Kernel smut',
    'Head Smut',
    'Rust',
    'loose smut'
]

num_classes = len(jowar_classes)

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# Freeze pretrained layers
base_model.trainable = False

jowar_model_v2 = models.Sequential([
    layers.Input(shape=(224, 224, 3)),

    # Correct preprocessing for MobileNetV2
    layers.Lambda(
        tf.keras.applications.mobilenet_v2.preprocess_input
    ),

    base_model,

    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),

    layers.Dense(
        num_classes,
        activation='softmax'
    )
])

jowar_model_v2.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

jowar_model_v2.summary()
```

```python
jowar_history_v2 = jowar_model_v2.fit(
    jowar_train_ds,
    validation_data=jowar_val_ds,
    epochs=5
)
```
