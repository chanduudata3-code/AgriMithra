```python
import os

RICE_DATASET_PATH = r"D:\AgriMithra datasets\archive (1)\Rice_Leaf_AUG"

print("Dataset exists:", os.path.exists(RICE_DATASET_PATH))

print("\nRice classes:")
for folder in sorted(os.listdir(RICE_DATASET_PATH)):
    print(folder)
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    RICE_DATASET_PATH,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    RICE_DATASET_PATH,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

rice_class_names = train_ds.class_names

print("\nClasses:", rice_class_names)
print("Number of classes:", len(rice_class_names))
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.cache().shuffle(1000).prefetch(
    buffer_size=AUTOTUNE
)

val_ds = val_ds.cache().prefetch(
    buffer_size=AUTOTUNE
)

print("Dataset optimization completed!")
```

```python
rice_class_names = [
    'Bacterial Leaf Blight',
    'Brown Spot',
    'Healthy Rice Leaf',
    'Leaf Blast',
    'Leaf scald',
    'Sheath Blight'
]

print(rice_class_names)
print("Number of classes:", len(rice_class_names))
```

```python
from tensorflow.keras import layers, models

num_classes = 6

rice_model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),

    layers.Rescaling(1./255),

    layers.Conv2D(32, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(64, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(128, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.GlobalAveragePooling2D(),

    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(num_classes, activation='softmax')
])

rice_model.summary()
```

```python
rice_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history = rice_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)
```

```python
import os

RICE_MODEL_PATH = r"D:\AgriMithra datasets\rice_model.keras"

rice_model.save(RICE_MODEL_PATH)

print("Rice model saved successfully!")
print("Saved at:", RICE_MODEL_PATH)
```

```python
val_loss, val_accuracy = rice_model.evaluate(val_ds)

print("Validation Loss:", round(val_loss, 4))
print("Validation Accuracy:", round(val_accuracy * 100, 2), "%")
```
